```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import sklearn

print(sklearn.__version__)
```

```python
import pandas as pd 

heart_disease = pd.read_csv("./data/heart-disease.csv")
heart_disease
```

```python
X = heart_disease.drop("target", axis=1)
y = heart_disease["target"]
X.head()
```

```python
y.head() , y.value_counts()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25)

X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.ensemble import RandomForestClassifier
clf = RandomForestClassifier()
clf.get_params()
```

```python
clf.fit(X_train, y_train)
```

```python
y_label = clf.predict(np.array([0,2,3,4]))
```

```python

y_preds = clf.predict(X_test)

```

```python
train_acc = clf.score(X_train, y_train)
test_acc = clf.score(X_test, y_test)
print(f"Train acc: {train_acc*100:.2f}%")
print(f"Test acc: {test_acc*100:.2f}% ")
```

```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

print(classification_report(y_test, y_preds))
```

```python
conf_mat = confusion_matrix(y_test, y_preds)
conf_mat
```

```python
accuracy_score(y_test, y_preds)
```

```python
np.random.seed(42)
for i  in range(100,500,10):
    print(f"Trying model with {i} estimators ...")
    model = RandomForestClassifier(n_estimators=i).fit(X_train,y_train)
    print(f"Model accuracy on test set {i}: {model.score(X_test,y_test)*100:.2f}%")
```

```python
from sklearn.model_selection import cross_val_score

np.random.seed(42)
for i in range (100,500,10):
    print(f"Trying Model with {i} estimator...")
    model = RandomForestClassifier(n_estimators=i).fit(X_train,y_train)


    model_score = model.score(X_test,y_test)
    print(f"MOdel accuracy on single test set split: {model_score*100:.2f}%")

    cross_val_mean = np.mean(cross_val_score(model,X,y,cv=5))
    print(f"5-fold cross-validation score : {cross_val_mean*100:.2f}%")

    print("")


```

```python
np.random.seed(42)

from sklearn.model_selection import GridSearchCV

param_grid = {'n_estimators': [i for i in range(100,500,10)]}

grid= GridSearchCV(estimator=RandomForestClassifier(), param_grid=param_grid,cv=5,verbose =1)


grid.fit(X,y)
print(f"the best parameter values are : {grid.best_params_}")
print(f"WIth a score of :{grid.best_score_*100:.2f}%")
```

```python
clf = grid.best_estimator_
clf
```
